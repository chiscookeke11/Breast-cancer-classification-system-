# Importing the dependencies 

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import  train_test_split
from sklearn.linear_model import  LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import  StandardScaler
```

```python
# Loading the dataset
cancer_data = pd.read_csv("./sample_data/data.csv")
```

# Data Preprocessing 


```python
# Checking the first 5 rows

cancer_data.head()
```

```python
# Checking the last 5 rows

cancer_data.tail()
```

```python
# Checkin the number of rows and columns

cancer_data.shape
```

```python
# Checking for more information on the data

cancer_data.info()
```

```python
# Checking for null values

cancer_data.isnull().sum()
```

```python
# Dropping the "unnamed" column

cancer_data.drop(columns=['Unnamed: 32', 'id'] , axis=1, inplace=True)
```

```python
cancer_data.head()
```

```python
# Encoding

#  The diagnosis has values; M & B.
# M - Malignant
# B - Benigh

# We will replace M with 0 and B with 1



cancer_data["diagnosis"] = cancer_data["diagnosis"].replace({
    "M": 0,
    "B": 1
})
```

```python
cancer_data.head()
```

```python
# Getting the classification of the diagnosis column

cancer_data["diagnosis"].value_counts()
```

```python
# Checking the statistical measures of the dataset

cancer_data.describe()
```

# Seperating the data into features and target 


```python
X = cancer_data.drop(columns='diagnosis', axis=1)
Y = cancer_data['diagnosis']
```

```python
X.head()
```

```python
Y.head()
```

# Splitting the data into train and test data 


```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

# Scaling the dataset

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# Model Training 


#### Logistic regression 

```python
model = LogisticRegression(max_iter=1000)
```

```python
# Training the Logistic regression model using the training data

model.fit(X_train_scaled, Y_train)
```

# Making data predictions

```python
#  Prediction on training data

training_data_prediction = model.predict(X_train_scaled)

accuracy_score_of_training_data = accuracy_score(Y_train, training_data_prediction)

print("Prediction on training data", accuracy_score_of_training_data)
```

```python
#  Prediction on test data

test_data_prediction = model.predict(X_test_scaled)

accuracy_score_of_test_data = accuracy_score(Y_test, test_data_prediction)

print("Prediction on testing data:", accuracy_score_of_test_data)
```
